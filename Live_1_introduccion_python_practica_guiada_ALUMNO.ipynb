## 0.6. Funciones

El ROI (Retorno de la Inversión) es una métrica financiera que mide el beneficio o la ganancia obtenida en relación con el dinero gastado en una inversión. Sirve para saber si una acción, campaña o negocio es rentable.

- La fórmula básica es: `(Ingresos - Costes) / Costes x 100`.
- El resultado se expresa en porcentaje.
- Un porcentaje positivo muestra ganancia.
- Un porcentaje negativo muestra pérdida



````python
def calcular_roi(beneficio, coste):
    if coste == 0:
        return None

    return beneficio / coste


roi = calcular_roi(3200, 10000)
print(f"ROI: {roi} y {roi:.2%}")
````

### ✏️ 3.4. Ecommerce — resumir una cesta


Crea una función que reciba una lista de precios y devuelva:
- total;
- media;
- máximo.

````python
def resumir_cesta(precios):
    total = sum(precios)
    media = total / len(precios)
    maximo = max(precios)

    return total, media, maximo
````

### ✏️ 3.5. Aplicar una función a varios registros

### ¿Qué tareas empezarían a ser incómodas si tuviéramos 100.000 registros?

Con 100.000 registros empezarían a resultar incómodas tareas como:

* recorrer manualmente todos los registros

* calcular medias, máximos o mínimos
* crear nuevas variables registro por registro
* agrupar por categoría utilizando diccionarios y acumuladores
* gestionar valores ausentes
* buscar duplicados
* combinar distintos conjuntos de datos
* inspeccionar los resultados con `print()`
* repetir muchas transformaciones sucesivas

Por ejemplo, ahora hemos hecho algo parecido a:

```python

for registro in registros:

    if canal not in ventas_por_canal:

    ventas_por_canal[canal] += registro["ventas"]
```

Más adelante veremos que Pandas permite expresar esta idea de una forma mucho más directa, conceptualmente parecida a:
